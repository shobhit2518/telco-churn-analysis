```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = pd.read_csv("df_cleaned")
df
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='InternetService',hue='Churn', data=df)
plt.title("Customer Churn by internet service type")
plt.xlabel("Internet Service")
plt.ylabel("Churn")
plt.show()
```

<h1>Payment method vs churn

```python
plt.figure(figsize=(10,5))
sns.countplot(x='PaymentMethod', hue="Churn", data = df)
plt.title("Payment method vs churn")
plt.xlabel("Payment Method")
plt.ylabel("Churn")
plt.show()
```

<h1>Chart 3: Senior Citizen vs Churn

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='SeniorCitizen', hue='Churn', data=df)
plt.title('Churn Distribution by Senior Citizen Status')
plt.xlabel('Senior Citizen (0: No, 1: Yes)')
plt.ylabel('Count')
plt.savefig('senior_citizen_churn.png')
plt.show()
```

<h1>Monthly Charges Distribution (KDE Plot)</h1>

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data=df, x='MonthlyCharges', hue='Churn', fill=True, common_norm=False)
plt.title('Density Distribution of Monthly Charges by Churn')
plt.xlabel('Monthly Charges ($)')

plt.show()
```

<h1>Tenure Boxplot with Churn

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='Churn', y='tenure', data=df)
plt.title('Tenure Distribution across Churn Status')
plt.xlabel('Churn Status')
plt.ylabel('Tenure (Months)')
plt.show()
```

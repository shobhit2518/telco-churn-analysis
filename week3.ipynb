```python
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('df_cleaned.csv')
```

```python
churned_charges = df[df['Churn'] == 'Yes']['MonthlyCharges']
not_churned_charges = df[df['Churn'] == 'No']['MonthlyCharges']
```

```python
t_stat, p_value = stats.ttest_ind(churned_charges, not_churned_charges, equal_var=False)
```

```python
print("--- Independent T-Test Results ---")
print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4e}")
```

```python
if p_value < 0.05:
    print("Result: Reject Null Hypothesis (Significant difference exists in Monthly Charges).\n")
else:
    print("Result: Fail to Reject Null Hypothesis.\n")
```

```python
contingency_table = pd.crosstab(df['Contract'], df['Churn'])
chi2, p_chi2, dof, ex = stats.chi2_contingency(contingency_table)
```

```python
print("--- Chi-Square Test Results ---")
print(f"Chi2 Statistic: {chi2:.4f}")
print(f"P-Value: {p_chi2:.4e}")
```

```python
if p_chi2 < 0.05:
    print("Result: Reject Null Hypothesis (Contract type significantly affects Churn).")
else:
    print("Result: Fail to Reject Null Hypothesis.")
```
